## Select Dataset Columns using Pandas `loc` and `iloc`
**Student:** Ayush Rajput

```python
import pandas as pd
df = pd.read_csv('iris_sample.csv')
df
```

## Using `loc` – Label Based Selection

```python
df.loc[:, 'species']
df.loc[:, ['sepal_length', 'petal_length', 'species']]
df.loc[0:3, ['id', 'sepal_length', 'species']]
```

## Using `iloc` – Integer Position Based Selection

```python
df.iloc[0:3, 0:3]
df.iloc[2:6, 1:5]
df.iloc[0, 5]
```

## Conclusion
`loc` selects using labels, while `iloc` selects using integer positions.
